```python

import pandas as pd
import numpy as np

df_raw = pd.read_csv('/content/reporte_cruceros_revenue_management.csv')

print("Dimensiones iniciales:", df_raw.shape)
df_raw.head()
```

```python
# ==========================================
# EXPLORACIÓN Y LIMPIEZA DE DATOS (EDA)
# ==========================================
df = df_raw.copy()

# arregla columnas de fecha
df['Fecha_Viaje'] = pd.to_datetime(df['Fecha_Viaje'])
df['Fecha_Reserva'] = pd.to_datetime(df['Fecha_Reserva'])

# elimina duplicados
df = df.drop_duplicates()

# elimina nulos
df = df.dropna()

print("Estructura después de limpiar:")
print(df.info())


```

```python
# Exportar a CSV
df.to_csv('cruceros_procesado.csv', index=False)
print("'cruceros_procesado.csv' ok")
```
